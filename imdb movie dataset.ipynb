```python
data['Meta_score']=pd.to_numeric(data['Meta_score'],errors='coerce')
```

```python
data['No_of_Votes']=pd.to_numeric(data['No_of_Votes'],errors='coerce')
```

```python
top_movies=data.sort_values(by='IMDB_Rating',ascending=False).head(10)
print(top_movies[['Series_Title','IMDB_Rating']])
```

```python
top_directors = data['Director'].value_counts().head(5)

plt.figure()
plt.pie(
    top_directors.values,
    labels=top_directors.index,
    autopct='%1.1f%%'
)
plt.title("Top 5 Directors")
plt.show()
```
